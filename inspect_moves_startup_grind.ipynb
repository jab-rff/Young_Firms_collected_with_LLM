```python
import polars as pl
import re

def clean_firm_name(s):
    if s is None:
        return None
    s = str(s).lower().strip()
    s = re.sub(r"[()]", " ", s)
    s = re.sub(r"[^a-z0-9&/+ -]", "", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s

def find_firms_in_df(df: pl.DataFrame, firm_col: str, firm_list: list[str]) -> pl.DataFrame:
    firm_list_clean = [clean_firm_name(f) for f in firm_list]

    return (
        df.with_columns(
            pl.col(firm_col)
            .map_elements(clean_firm_name, return_dtype=pl.Utf8)
            .alias("_firm_clean")
        )
        .filter(pl.col("_firm_clean").is_in(firm_list_clean))
    )

def find_firms_across_dfs(
    firm_list: list[str],
    df2: pl.DataFrame,
    df3: pl.DataFrame,
    df4: pl.DataFrame,
    df5: pl.DataFrame,
    df6: pl.DataFrame,
    df2_firm_col: str = "firm",
    df3_firm_col: str = "Firm",
    df4_firm_col: str = "firm",
    df5_firm_col: str = "firm",
    df6_firm_col: str = "firm",
):
    df2_firms_found = find_firms_in_df(df2, df2_firm_col, firm_list)
    df3_firms_found = find_firms_in_df(df3, df3_firm_col, firm_list)
    df4_firms_found = find_firms_in_df(df4, df4_firm_col, firm_list)
    df5_firms_found = find_firms_in_df(df5, df5_firm_col, firm_list)
    df6_firms_found = find_firms_in_df(df6, df6_firm_col, firm_list)

    return (
        df2_firms_found,
        df3_firms_found,
        df4_firms_found,
        df5_firms_found,
        df6_firms_found,
    )
```

```python
import polars as pl
import re

def clean_text(s):
    if s is None:
        return ""
    s = s.lower()
    s = re.sub(r"[^a-z0-9 ]", " ", s)
    s = re.sub(r"\s+", " ", s)
    return s

def find_firms_in_text(df: pl.DataFrame, firm_list: list[str], text_col: str = "text_into_model"):
    # clean firm list
    firm_list_clean = [clean_text(f) for f in firm_list]

    # clean text column once
    df = df.with_columns(
        pl.col(text_col)
        .map_elements(clean_text, return_dtype=pl.Utf8)
        .alias("_text_clean")
    )

    hits = []
    found_firms = set()

    for firm, firm_clean in zip(firm_list, firm_list_clean):
        mask = df["_text_clean"].str.contains(firm_clean)
        df_hit = df.filter(mask)

        if df_hit.height > 0:
            hits.append(df_hit.with_columns(pl.lit(firm).alias("matched_firm")))
            found_firms.add(firm)

    if hits:
        df_hits = pl.concat(hits)
    else:
        df_hits = pl.DataFrame()

    return df_hits, list(found_firms)
```
